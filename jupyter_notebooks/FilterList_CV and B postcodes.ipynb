```python
import duckdb

# 1. Generate Birmingham prefixes: B1% through B98%
birmingham_prefixes = [f"'B{i}%'" for i in range(1, 99)]

# 2. Generate Coventry prefixes: CV1% through CV47%
coventry_prefixes = [f"'CV{i}%'" for i in range(1, 48)]

# Combine them into a single list
all_prefixes = birmingham_prefixes + coventry_prefixes

# Build the SQL WHERE condition: "postcode LIKE 'B1%' OR postcode LIKE 'B2%' OR ..."
prefix_condition = " OR ".join([f"postcode LIKE {p}" for p in all_prefixes])

# 3. Define file paths
input_csv = "../dataset/epc.csv"
output_csv = "../dataset/midlands_subset.csv"

# 4. Construct and execute the DuckDB query
query = f"""
COPY (
    SELECT *
    FROM '{input_csv}'
    WHERE {prefix_condition}
) TO '{output_csv}' (HEADER, DELIMITER ',');
"""

print(f"Filtering dataset across {len(all_prefixes)} postcode patterns...")
duckdb.query(query)
print(f"Success! Midlands subset exported to: {output_csv}")
```
